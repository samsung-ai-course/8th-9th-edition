## Exercise 11: RAG (Retrieval-Augmented Generation) with Movies

**Goal**: Build a RAG system that answers questions about movies using retrieval-augmented generation.

**What is RAG?**
- **Retrieval**: Find relevant movie information using similarity search (TF-IDF from Exercise 5)
- **Augmentation**: Include retrieved context in the prompt
- **Generation**: Use an LLM to generate answers based on retrieved context

**Why RAG?**
- LLMs have knowledge cutoff dates and may not know about specific movies in our dataset
- RAG provides up-to-date, relevant context from our movie database
- Better than fine-tuning for domain-specific knowledge

**Your Task**: Implement a RAG system that:
1. Takes a user question about movies (e.g., "What are some good sci-fi movies?")
