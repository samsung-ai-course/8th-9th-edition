# Team Generator

This notebook reads student names from a text file and forms teams of 3.

**Rules:**
- If remainder is 1: make one team of 4
- If remainder is 2: make two teams of 4
- Otherwise: all teams of 3

```python
def form_teams(filename):
    """
    Read student names from a text file and form teams.
    
    Args:
        filename: Path to text file with student names (one per line)
    
    Returns:
        List of tuples (team_name, team_members)
    """
    # Read student names from file
    with open(filename, 'r') as f:
        students = [line.strip() for line in f if line.strip()]
    
    num_students = len(students)
    remainder = num_students % 3
    
    teams = []
    student_index = 0
    team_letter = 65  # ASCII code for 'A'
    
    if remainder == 1:
        # Make one team of 4, rest teams of 3
        # First team gets 4 members
        team_name = f"Team {chr(team_letter)}"
        team_members = students[student_index:student_index + 4]
        teams.append((team_name, team_members))
        student_index += 4
        team_letter += 1
        
        # Rest are teams of 3
        while student_index < num_students:
            team_name = f"Team {chr(team_letter)}"
            team_members = students[student_index:student_index + 3]
            teams.append((team_name, team_members))
            student_index += 3
            team_letter += 1
    
    elif remainder == 2:
        # Make two teams of 4, rest teams of 3
        # First two teams get 4 members each
        for _ in range(2):
            team_name = f"Team {chr(team_letter)}"
            team_members = students[student_index:student_index + 4]
            teams.append((team_name, team_members))
            student_index += 4
            team_letter += 1
        
        # Rest are teams of 3
        while student_index < num_students:
            team_name = f"Team {chr(team_letter)}"
            team_members = students[student_index:student_index + 3]
            teams.append((team_name, team_members))
            student_index += 3
            team_letter += 1
    
    else:
        # remainder == 0, all teams of 3
        while student_index < num_students:
            team_name = f"Team {chr(team_letter)}"
            team_members = students[student_index:student_index + 3]
            teams.append((team_name, team_members))
            student_index += 3
            team_letter += 1
    
    return teams
```

```python
def print_teams(teams):
    """
    Print teams in a formatted way.
    
    Args:
        teams: List of tuples (team_name, team_members)
    """
    print("=" * 50)
    print("TEAM ASSIGNMENTS")
    print("=" * 50)
    print()
    
    for team_name, members in teams:
        print(f"{team_name}:")
        for member in members:
            print(f"  - {member}")
        print()
    
    print("=" * 50)
    print(f"Total teams: {len(teams)}")
    print(f"Total students: {sum(len(members) for _, members in teams)}")
```
