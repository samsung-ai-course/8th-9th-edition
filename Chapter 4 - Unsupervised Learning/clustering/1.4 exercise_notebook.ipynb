# Vis functions
This functions help you to visualize the points in the map.
 - `visualize_points` will help you visualize all the points while you still dont have a `cluster` column with the cluster labels.  
 - `visualize_clusters` will show you clusters with different colors on the map.  
 - `visualize_points_with_continuous_color` will show you points on the map colorful with a continuous colour

```python

def visualize_points(df, lat_col='latitude', lon_col='longitude', point_color='blue', zoom=10):
    """
    Visualize points on a map with a single color.

    Parameters:
    - df (pd.DataFrame): The DataFrame containing the data.
    - lat_col (str): Name of the column for latitude.
    - lon_col (str): Name of the column for longitude.
    - point_color (str): Color for the points.
    - zoom (int): Initial zoom level for the map.

    Returns:
    - folium.Map: A folium map object with points visualized.
    """
    # Create a base map centered at the mean latitude and longitude
    map_center = [df[lat_col].mean(), df[lon_col].mean()]
    map_points = folium.Map(location=map_center, zoom_start=zoom)

    # Add scatter points (circle markers) with a single color
    for _, row in df.iterrows():
        folium.CircleMarker(
            location=[row[lat_col], row[lon_col]],
            radius=1,  # Size of the circle
            color=point_color,  # Outline color
            fill=True,
            fill_color=point_color,  # Fill color
            fill_opacity=0.7,  # Opacity of the fill
            popup=f"Latitude: {row[lat_col]}, Longitude: {row[lon_col]}"
        ).add_to(map_points)

    # Display the map inline
    display(map_points)
    return 0

def visualize_clusters(df, lat_col='latitude', lon_col='longitude', cluster_col='cluster', zoom=10):
    """
    Visualize clustered points on a map with a color per cluster.

    Parameters:
    - df (pd.DataFrame): The DataFrame containing the data.
    - lat_col (str): Name of the column for latitude.
    - lon_col (str): Name of the column for longitude.
    - cluster_col (str): Name of the column for cluster labels.
    - zoom (int): Initial zoom level for the map.

    Returns:
    - folium.Map: A folium map object with clusters visualized.
    """
    # Generate a color map for clusters
    unique_clusters = df[cluster_col].unique()
    colors = cm.tab10.colors  # Get a palette with 10 distinct colors
    cluster_colors = {
        cluster: f'#{int(r*255):02x}{int(g*255):02x}{int(b*255):02x}' 
        for cluster, (r, g, b) in zip(unique_clusters, colors)
    }
    cluster_colors[-1] = 'gray'  # Mark noise (-1 cluster) as gray

    # Create the map centered at the mean latitude and longitude
    map_center = [df[lat_col].mean(), df[lon_col].mean()]
    map_points = folium.Map(location=map_center, zoom_start=zoom)

    # Add scatter points (circle markers) colored by cluster
    for _, row in df.iterrows():
        cluster = row[cluster_col]
        color = cluster_colors.get(cluster, 'gray')  # Default to gray for noise
        folium.CircleMarker(
            location=[row[lat_col], row[lon_col]],
            radius=5,
            color=color,
            fill=True,
            fill_color=color,
            fill_opacity=0.7,
            popup=f"Cluster: {cluster}, Latitude: {row[lat_col]}, Longitude: {row[lon_col]}"
        ).add_to(map_points)

    # Display the map inline
    display(map_points)
    return 0



def visualize_points_with_continuous_color(df, lat_col='latitude', lon_col='longitude', value_col=None, zoom=10):
    """
    Visualize points on a folium map with colors representing a continuous variable.

    Parameters:
    - df (pd.DataFrame): The DataFrame containing the data.
    - lat_col (str): Name of the column for latitude.
    - lon_col (str): Name of the column for longitude.
    - value_col (str): Name of the column with continuous values for color mapping.
    - zoom (int): Initial zoom level for the map.

    Returns:
    - folium.Map: A folium map object with points visualized.
    """
    if value_col is None:
        raise ValueError("A column for continuous variable must be specified as value_col.")

    # Create a base map centered at the mean latitude and longitude
    map_center = [df[lat_col].mean(), df[lon_col].mean()]
    map_points = folium.Map(location=map_center, zoom_start=zoom)

    # Create a colormap using Folium's LinearColormap
    min_value, max_value = df[value_col].min(), df[value_col].max()
    colormap = linear.RdYlBu_09.scale(min_value, max_value)  # 'YlOrRd' for warm colors
    

    # Add scatter points (circle markers) with continuous color
    for _, row in df.iterrows():
        value = row[value_col]
        color = colormap(value)  # Get the color for the value
        folium.CircleMarker(
            location=[row[lat_col], row[lon_col]],
            radius=1,
            color=color,  # Outline color
            fill=True,
            fill_color=color,  # Fill color
            fill_opacity=1,  # Opacity of the fill
            popup=f"{value_col}: {value}, Latitude: {row[lat_col]}, Longitude: {row[lon_col]}"
        ).add_to(map_points)

    # Add the color legend to the map
    colormap.add_to(map_points)

    # Display the map inline
    display(map_points)
    return 0
```
