## 📚 Python Classes - Quick Reference

**Before Exercise 12**, you'll need to understand **Python classes**. If you're new to classes, here's a quick overview:

### What is a Python Class?

A **class** is like a blueprint for creating objects. It groups related data (attributes) and functions (methods) together.

**Key Concepts:**
- **`class`**: Keyword to define a class
- **`__init__`**: Special method that runs when you create an object (constructor)
- **`self`**: Reference to the current object instance
- **Methods**: Functions defined inside a class

### Simple Example:

```python
class Calculator:
    def __init__(self, name):
        """Initialize the calculator with a name"""
        self.name = name  # Store data in the object
        self.history = []  # Another piece of data
    
    def add(self, a, b):
        """Method to add two numbers"""
        result = a + b
        self.history.append(f"{a} + {b} = {result}")
        return result
    
    def get_history(self):
        """Method to get calculation history"""
        return self.history

# Using the class:
calc = Calculator("My Calculator")  # Create an object
result = calc.add(5, 3)  # Call a method
print(calc.get_history())  # Access another method
```
